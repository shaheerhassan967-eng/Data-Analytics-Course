```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("ecommerce_dataset.csv")
df.head()
```

```python
df.dtypes
```

```python
df = df.drop(["Latitude", "Longtitude"], axis=1)
df
```

```python
df = df.rename(columns={"City Code": "City","Order Priority Code": "Order Priority"})
df
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df
```

```python
df.isnull().sum()
```

```python
df["total"] = df["total"].fillna(df["total"].median())
df
```

```python
top_products = df.groupby("product")["quantity"].sum().sort_values(ascending=False).head(5)
top_products
```

```python
sns.barplot(x=top_products.values, y=top_products.index)
plt.title("Top 5 Best Selling Products")
plt.xlabel("Quantity Sold")
plt.ylabel("Product")
plt.show()
```

Watch is the best-selling product.
Mobile is the second best-selling product.
Headphones is third.
Bag and Shoes follow.
Watch sold 256 units, making it the top-selling product.

```python
payment = df["payment_method"].value_counts()
payment
```

```python
plt.pie(payment.values, labels=payment.index, autopct="%1.1f%%")
plt.title("Payment Method Distribution")
plt.show()
```

JazzCash has the highest number of orders.
Credit Card and Easypaisa are also commonly used.
Debit Card has the lowest number of orders.
Overall, digital payment methods are widely used in the dataset.

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]
```

```python
monthly_sales = df.groupby("Month")["total"].sum().reindex(month_order)
monthly_sales
```

```python
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")
plt.title("Month-wise Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

March has the highest sales.
January and April also show strong sales.
September has the lowest sales.
Sales are generally higher during the first half of the year compared with several later months.

```python
df["City"] = df["City"].replace({1: "Karachi",2: "Lahore",3: "Islamabad",4: "Faisalabad",5: "Quetta",6: "Peshawar"})
df
```

```python
df["City"].value_counts()
```

```python
city_sales = df.groupby("City")["total"].sum().sort_values()
sns.barplot(x=city_sales.values, y=city_sales.index)
plt.title("City-wise Sales")
plt.xlabel("Sales")
plt.ylabel("City")
plt.show()
```

Lahore has the highest sales among all cities.
Quetta is the second-highest.
Karachi has the lowest sales.
Lahore contributes significantly more sales compared with the other cities.

```python
category_sales = df.groupby(["Month", "category"])["total"].sum().unstack()
category_sales
```

```python
category_sales = category_sales.reindex(month_order)
category_sales
```

```python
category_sales.plot(kind="bar")
plt.title("Monthly Sales by Category")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.legend(title="Category")
plt.show()
```
