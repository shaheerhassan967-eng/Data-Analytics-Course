```python
student_data = data[1:]
```

```python
student_data
```

```python
for Gr in student_data:

    percentage = float(Gr[5])

    if percentage >= 80:
        Grade = 'A'
    elif 70 <= percentage < 80:
        Grade = 'B'
    elif 60 <= percentage < 70:
        Grade = 'C'
    elif 50 <= percentage < 60:
        Grade = 'D'
    else:
        Grade = 'F'

    Gr.append(Grade)
    print(Grade)
```
